```python
from pathlib import Path
import sys
root = Path.cwd()
if root.name == "model":
    root = root.parent
sys.path.insert(0, str(root))
import torch
from model.dino import DINO
```

```python
model = DINO(
    shape_in=(1, 1, 32, 32), spatial_hidden_dim=16,
    temporal_hidden_dim=32, output_channels=1,
    num_spatial_layers=4, num_temporal_layers=8,
    moment_constraint=False, grid_scaling=False,
).eval()
with torch.no_grad():
    prediction = model(torch.randn(1, 1, 1, 32, 32))
print(prediction.shape)
```
